```python
# Crear una copia para evitar errores
df_limpio = df_limpio.copy()

# Limpiar saltos de línea y espacios extras
df_limpio['Ubicación_limpia'] = df_limpio['Ubicación'].astype(str)
df_limpio['Ubicación_limpia'] = df_limpio['Ubicación_limpia'].str.replace('\n', ' ', regex=False)
df_limpio['Ubicación_limpia'] = df_limpio['Ubicación_limpia'].str.replace(r'\s+', ' ', regex=True)
df_limpio['Ubicación_limpia'] = df_limpio['Ubicación_limpia'].str.strip()

# Mostrar resultado
df_limpio[['Ubicación', 'Ubicación_limpia']].head()
```

```python
# Función para extraer zona
def obtener_zona(texto):
    partes = texto.split(',')

    if len(partes) >= 2:
        zona = partes[-2].strip()
    else:
        zona = texto.strip()

    zona = zona.replace('Casa en Venta', '').strip()

    if zona == '':
        zona = 'Toluca De Lerdo, México'

    return zona

# Crear nueva variable Zona
df_limpio['Zona'] = df_limpio['Ubicación_limpia'].apply(obtener_zona)

# Revisar resultado
df_limpio[['Ubicación_limpia', 'Zona']].head(15)
```

```python
# Contar las zonas más frecuentes
zonas_mas_frecuentes = df_limpio['Zona'].value_counts().head(10)

plt.figure(figsize=(9, 5), dpi=100)

plt.barh(
    zonas_mas_frecuentes.index,
    zonas_mas_frecuentes.values,
    color='BLUE'
)

plt.xlabel("Cantidad de casas")
plt.ylabel("Zona")
plt.title("Zonas con mayor número de casas registradas")

plt.gca().invert_yaxis()

plt.show()
```

```python
import matplotlib.pyplot as plt

# Create tabla_zonas by grouping df_limpio by 'Zona' and calculating count and mean of 'Precio'
tabla_zonas = df_limpio.groupby('Zona').agg(
    cantidad=('Precio', 'count'),
    precio_promedio=('Precio', 'mean')
)

# Seleccionar las zonas con al menos 5 registros
tabla_zonas_filtrada = tabla_zonas[tabla_zonas['cantidad'] >= 5]

# Tomar las 10 zonas con mayor precio promedio
# Sort by 'precio_promedio' in descending order to get the highest average prices
zonas_precio_alto = tabla_zonas_filtrada.sort_values(by='precio_promedio', ascending=False).head(10)

plt.figure(figsize=(10, 5), dpi=100)

plt.barh(
    zonas_precio_alto.index,
    zonas_precio_alto['precio_promedio'],
    color='BLUE'
)

plt.xlabel("Precio promedio")
plt.ylabel("Zona")
plt.title("Zonas con mayor precio promedio")

plt.gca().invert_yaxis()

plt.show()
```

```python
import statsmodels.api as sm
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import matplotlib.pyplot as plt

# Perform ANOVA to check for significant differences between groups (Zones)
# C(Zona) tells statsmodels that 'Zona' should be treated as a categorical variable
model = sm.formula.ols('Precio ~ C(Zona)', data=df_limpio).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
# print(anova_table) # Uncomment to see ANOVA results

# Perform Tukey HSD post-hoc test
tukey = pairwise_tukeyhsd(endog=df_limpio['Precio'], # Dependent variable
                          groups=df_limpio['Zona'],   # Groups (independent variable)
                          alpha=0.05)                 # Significance level

# Gráfico de resultados de Tukey
tukey.plot_simultaneous()

plt.xlabel("Precio")
plt.ylabel("Zona")
plt.title("Comparación de precios promedio entre zonas")

plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# Re-calculate promedio_zona and codigos_zona as df_limpio might have been reset
promedio_zona = df_limpio.groupby('Zona')['Precio'].mean().sort_values()
codigos_zona = {}
for i, zona in enumerate(promedio_zona.index):
    codigos_zona[zona] = i

# Add 'Zona_cod' to df_limpio
df_limpio['Zona_cod'] = df_limpio['Zona'].map(codigos_zona)

# Define features (X) and target (Y) using the columns previously established for correlation
X = df_limpio[['Area', 'Baños', 'Habitaciones', 'Zona_cod']]
Y = df_limpio['Precio']

# Split data into training and testing sets (using a fixed random_state for reproducibility)
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

# Initialize and train the Linear Regression model
modelo = LinearRegression()
modelo.fit(X_train, Y_train)

# Generate predictions for the full dataset (X) as implied by the original deleted cell
predicciones = modelo.predict(X)

comparacion = pd.DataFrame()

comparacion['Precio_real'] = Y
comparacion['Precio_estimado'] = predicciones
comparacion['Error'] = comparacion['Precio_real'] - comparacion['Precio_estimado']

comparacion.head()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(Y, predicciones)
rmse = np.sqrt(mean_squared_error(Y, predicciones))
r2 = r2_score(Y, predicciones)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```
